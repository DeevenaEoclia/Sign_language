# 1. What is a video?

A **video** is just a fast slideshow of still images.
Each still image is called a **frame**.

**FPS (Frames Per Second)** tells you how many frames are shown every second.
If FPS = 30, then 1 second of video = 30 separate photos shown quickly, one after another.

```text
Video
 ↓
Frame 1
Frame 2
Frame 3
...
Frame N
```

**Tenglish:** "Video ante konni photos ni fast ga oka tarvatha okati chupించడం. Video lo prathi photo ni 'frame' antaru. FPS ante prathi second ki entha frames vastundi ani."

OpenCV (`cv2`) is the Python library we use to read videos frame by frame.

````python
import cv2
import matplotlib.pyplot as plt

print("OpenCV version:", cv2.__version__)
````

## 2. Pick a sample ISL video

````python
video_path = None
for root, dirs, files in os.walk(VIDEOS_DIR):
    for f in files:
        if f.lower().endswith((".mp4", ".avi", ".mov")):
            video_path = os.path.join(root, f)
            break
    if video_path:
        break

if video_path is None:
    print("No video found yet. Add a video under data/videos/<WORD>/ and re-run this cell.")
else:
    print("Using video:", video_path)
````

## 3. Open the video and read its basic properties

````python
if video_path:
    cap = cv2.VideoCapture(video_path)

    width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
    height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
    fps = cap.get(cv2.CAP_PROP_FPS)
    frame_count = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
    duration = frame_count / fps if fps > 0 else 0

    print(f"Width:  {width} px")
    print(f"Height: {height} px")
    print(f"FPS:    {fps}")
    print(f"Total frames: {frame_count}")
    print(f"Duration: {duration:.2f} seconds")

    cap.release()
````

## 4. Extract frames one by one

````python
def extract_frames(path):
    cap = cv2.VideoCapture(path)
    frames = []
    while True:
        success, frame = cap.read()
        if not success:
            break
        frames.append(frame)
    cap.release()
    return frames

if video_path:
    frames = extract_frames(video_path)
    print(f"Extracted {len(frames)} frames.")
    print("Shape of one frame (height, width, channels):", frames[0].shape)
````

## 5. Display a few frames

````python
if video_path:
    num_to_show = min(5, len(frames))
    step = max(1, len(frames) // num_to_show)

    fig, axes = plt.subplots(1, num_to_show, figsize=(15, 4))
    for i in range(num_to_show):
        idx = i * step
        rgb_frame = cv2.cvtColor(frames[idx], cv2.COLOR_BGR2RGB)
        axes[i].imshow(rgb_frame)
        axes[i].set_title(f"Frame {idx}")
        axes[i].axis("off")
    plt.tight_layout()
    plt.show()
````
